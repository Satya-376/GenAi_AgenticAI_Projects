```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
# 1. Set your OpenAI API key
os.environ["OPENAI_API_KEY"] = "Enter your OpenAi key here"
```

```python
# 2. Initialize the OpenAI model
llm = ChatOpenAI(model="gpt-6-luna")
```

```python
# 3. Create a dynamic translation prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful translator."),
    ("user", "Translate the following text into {target_language}: {text}")
])
```

```python

# 4. Build the chain
translation_chain = prompt | llm
```

```python
# 5. Function to translate text into any language
def translate_text(text, target_language):
    result = translation_chain.invoke({
        "text": text,
        "target_language": target_language
    })
    return result.content
```

```python
# Example usage
print("English → Hindi:", translate_text("Good morning, have a nice day!", "Hindi"))
print("English → Spanish:", translate_text("Good morning, have a nice day!", "Spanish"))
print("English → German:", translate_text("Good morning, have a nice day!", "German"))
```
