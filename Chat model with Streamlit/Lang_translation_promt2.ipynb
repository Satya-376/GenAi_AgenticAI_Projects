```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
# 1. Set your OpenAI API key
os.environ["OPENAI_API_KEY"] = "Enter your OpenAI key here"
```

```python
# 2. Initialize the OpenAI model
llm = ChatOpenAI(model="gpt-6-luna")
```

```python
# 3. Create a translation prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful translator. Translate the given text into French."),
    ("user", "{text}")
])
```

```python
# 4. Build the chain
translation_chain = prompt | llm

# 5. Run the chain with input text
result = translation_chain.invoke({"text": "Good morning, have a nice day!"})

print("Translated Text:", result.content)
```
